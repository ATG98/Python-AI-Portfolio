<b>Problem 1. Fashion MNIST Dataset</b><br>
Build a classification model for Fashion MNIST dataset. Fashion MNIST is a dataset of Zalando's article images—consisting of a training set of 60,000 examples and a test set of 10,000 examples. Each example is a 28x28 grayscale image, associated with a label from 10 classes. To load the dataset, you may want to use `tf.keras.datasets.fashion_mnist`. Also write codes to answer the following questions.<br>
(1) How many samples does the dataset have for training and test? What is the label distribution?<br>
(2) Show an example image for each class.<br>
(3) Build a MLP classification model and evalutation the classification performance for the test samples. Show the summary of your model.<br>
(4) Build a CNN classification model and evalutation the classification performance for the test samples. Show the summary of your model.



```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt

# 데이터 로드
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.fashion_mnist.load_data()

# (1) 샘플 개수 및 라벨 분포 확인
print(f"훈련 세트 개수: {len(x_train)}")
print(f"테스트 세트 개수: {len(x_test)}")

unique, counts = np.unique(y_train, return_counts=True)
print(f"훈련 세트 라벨 분포: {dict(zip(unique, counts))}")

# (2) 클래스별 예제 이미지 시각화
class_names = ['T-shirt/top', 'Trouser', 'Pullover', 'Dress', 'Coat',
               'Sandal', 'Shirt', 'Sneaker', 'Bag', 'Ankle boot']

plt.figure(figsize=(10, 4))
for i in range(10):
    plt.subplot(2, 5, i+1)
    idx = np.where(y_train == i)[0][0]
    plt.imshow(x_train[idx], cmap='gray')
    plt.title(class_names[i])
    plt.axis('off')
plt.show()

# 전처리 (정규화)
x_train, x_test = x_train / 255.0, x_test / 255.0

# (3) MLP 모델 구축
mlp_model = tf.keras.Sequential([
    tf.keras.layers.Flatten(input_shape=(28, 28)),
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dense(10, activation='softmax')
])

mlp_model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
mlp_model.summary()

mlp_model.fit(x_train, y_train, epochs=5, verbose=1)
loss, acc = mlp_model.evaluate(x_test, y_test, verbose=0)
print(f"\nMLP 테스트 정확도: {acc:.4f}")

# (4) CNN 모델 구축
# 데이터 차원 변경 (28, 28) -> (28, 28, 1)
x_train_cnn = x_train[..., np.newaxis]
x_test_cnn = x_test[..., np.newaxis]

cnn_model = tf.keras.Sequential([
    tf.keras.layers.Conv2D(32, (3,3), activation='relu', input_shape=(28, 28, 1)),
    tf.keras.layers.MaxPooling2D((2,2)),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(64, activation='relu'),
    tf.keras.layers.Dense(10, activation='softmax')
])

cnn_model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
cnn_model.summary()

cnn_model.fit(x_train_cnn, y_train, epochs=5, verbose=1)
loss, acc = cnn_model.evaluate(x_test_cnn, y_test, verbose=0)
print(f"\nCNN 테스트 정확도: {acc:.4f}")
```

<b>Problem 2. Cifar10 Dataset</b><br>
The CIFAR-10 dataset consists of 60,000 32x32 colour images in 10 classes, with 6,000 images per class. There are 50,000 training images and 10,000 test images. To load the dataset, you may want to use `tf.keras.datasets.cifar10`. Also write codes to answer the following questions.<br>
(1) How many samples does the dataset have for training and test? What is the label distribution?<br>
(2) What is the shape of a sample? Can you see why it is different from the shape of a sample in Problem 1? <br>
(2) Show an example image for each class. You may want to use `matplotlib.pyplot.imshow`.<br>
(3) Build a MLP classification model and evalutation the classification performance for the test samples. Show the summary of your model.<br>
(4) Build a CNN classification model and evalutation the classification performance for the test samples. Show the summary of your model.<br>

```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt

# 데이터 로드
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.cifar10.load_data()

# (1) 샘플 개수 및 라벨 분포 확인
print(f"훈련 데이터 개수: {x_train.shape[0]}")
print(f"테스트 데이터 개수: {x_test.shape[0]}")

unique, counts = np.unique(y_train, return_counts=True)
print(f"훈련 세트 라벨 분포: {dict(zip(unique, counts))}")

# (2) 샘플의 형태 (Shape) 확인 및 비교 설명
print(f"\n샘플의 형태(Shape): {x_train[0].shape}")
# 설명: Fashion MNIST는 (28, 28) 형태의 흑백(Gray) 이미지였으나,
# CIFAR-10은 (32, 32, 3) 형태의 컬러(RGB) 이미지입니다.
# 1) 해상도가 28x28에서 32x32로 커졌고,
# 2) 채널이 1개에서 3개(Red, Green, Blue)로 늘어났기 때문에 형태가 다릅니다.

# (2-2) 클래스별 예제 이미지 시각화
class_names = ['airplane', 'automobile', 'bird', 'cat', 'deer',
               'dog', 'frog', 'horse', 'ship', 'truck']

plt.figure(figsize=(10, 4))
for i in range(10):
    plt.subplot(2, 5, i+1)
    # i번 클래스에 해당하는 첫 번째 이미지 찾기
    idx = np.where(y_train.flatten() == i)[0][0]
    plt.imshow(x_train[idx])
    plt.title(class_names[i])
    plt.axis('off')
plt.show()

# 데이터 전처리 (정규화: 0~255 값을 0~1로 변경)
x_train, x_test = x_train / 255.0, x_test / 255.0

# (3) MLP 모델 구축
mlp_model = tf.keras.Sequential([
    tf.keras.layers.Flatten(input_shape=(32, 32, 3)), # 32x32x3 전체를 펼침
    tf.keras.layers.Dense(256, activation='relu'),
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dense(10, activation='softmax')
])

mlp_model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
mlp_model.summary()

mlp_model.fit(x_train, y_train, epochs=10, verbose=1)
mlp_model.evaluate(x_test, y_test, verbose=1)

# (4) CNN 모델 구축
cnn_model = tf.keras.Sequential([
    # 입력 형태: (32, 32, 3)
    tf.keras.layers.Conv2D(32, (3, 3), activation='relu', input_shape=(32, 32, 3)),
    tf.keras.layers.MaxPooling2D((2, 2)),
    tf.keras.layers.Conv2D(64, (3, 3), activation='relu'),
    tf.keras.layers.MaxPooling2D((2, 2)),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(64, activation='relu'),
    tf.keras.layers.Dense(10, activation='softmax')
])

cnn_model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
cnn_model.summary()

cnn_model.fit(x_train, y_train, epochs=10, verbose=1)
cnn_model.evaluate(x_test, y_test, verbose=1)
```
