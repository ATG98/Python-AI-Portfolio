<b>Problem 1. Newsgroup Classification</b><br>
Build a text classification model for newsgroup texts. There are 4 categories. Try both MLP and RNN models, and compare the accuracy and model size.


```python
from sklearn.datasets import fetch_20newsgroups
remove_list = ('headers', 'footers', 'quotes')
categories = ['alt.atheism', 'talk.politics.guns', 'comp.graphics', 'rec.autos']
newsgroups_train = fetch_20newsgroups(subset='train',remove=remove_list, categories=categories)
newsgroups_test = fetch_20newsgroups(subset='test',remove=remove_list, categories=categories)

x_train = newsgroups_train.data
y_train = newsgroups_train.target
x_test = newsgroups_test.data
y_test = newsgroups_test.target
labels = newsgroups_train.target_names
```

```python
# 데이터를 숫자로 변환
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

vocab_size = 5000
max_length = 200

tokenizer = Tokenizer(num_words=vocab_size, oov_token="<OOV>")
tokenizer.fit_on_texts(x_train)

x_train_padded = pad_sequences(tokenizer.texts_to_sequences(x_train), maxlen=max_length)
x_test_padded = pad_sequences(tokenizer.texts_to_sequences(x_test), maxlen=max_length)
```

```python
from tensorflow.keras import layers, models

mlp_model = models.Sequential([
    layers.Embedding(vocab_size, 16, input_length=max_length),
    layers.GlobalAveragePooling1D(),
    layers.Dense(24, activation='relu'),
    layers.Dense(4, activation='softmax')
])

mlp_model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
mlp_model.summary()
mlp_model.fit(x_train_padded, y_train, epochs=10, validation_data=(x_test_padded, y_test), verbose=1)
```

```python
rnn_model = models.Sequential([
    layers.Embedding(vocab_size, 16, input_length=max_length),
    layers.SimpleRNN(32),
    layers.Dense(4, activation='softmax')
])

rnn_model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
rnn_model.summary()
rnn_model.fit(x_train_padded, y_train, epochs=10, validation_data=(x_test_padded, y_test), verbose=1)
```

<b>Problem 2. Newsgroup Classification 2</b><br>
Repeat Problem 1 for all 20 categories. Just make `categories=None`. Try both MLP and RNN models,` and compare the performance and size.

```python
# 1. 20개 카테고리로 데이터 재로딩
newsgroups_train = fetch_20newsgroups(subset='train', remove=remove_list, categories=None)
newsgroups_test = fetch_20newsgroups(subset='test', remove=remove_list, categories=None)

x_train = newsgroups_train.data
y_train = newsgroups_train.target
x_test = newsgroups_test.data
y_test = newsgroups_test.target

# 2. 전처리 (이전 셀에서 정의한 tokenizer 그대로 사용)
x_train_padded = pad_sequences(tokenizer.texts_to_sequences(x_train), maxlen=max_length)
x_test_padded = pad_sequences(tokenizer.texts_to_sequences(x_test), maxlen=max_length)

# 3. MLP 모델 (출력층을 20으로 수정)
mlp_model_20 = models.Sequential([
    layers.Embedding(vocab_size, 16, input_length=max_length),
    layers.GlobalAveragePooling1D(),
    layers.Dense(24, activation='relu'),
    layers.Dense(20, activation='softmax') # 20개 카테고리
])
mlp_model_20.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
mlp_model_20.build(input_shape=(None, max_length))
print("--- MLP 모델 요약 ---")
mlp_model_20.summary()
mlp_model_20.fit(x_train_padded, y_train, epochs=10, validation_data=(x_test_padded, y_test), verbose=1)

# 4. RNN 모델 (출력층을 20으로 수정)
rnn_model_20 = models.Sequential([
    layers.Embedding(vocab_size, 16, input_length=max_length),
    layers.SimpleRNN(32),
    layers.Dense(20, activation='softmax') # 20개 카테고리
])
rnn_model_20.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
rnn_model_20.build(input_shape=(None, max_length))
print("\n--- RNN 모델 요약 ---")
rnn_model_20.summary()
rnn_model_20.fit(x_train_padded, y_train, epochs=10, validation_data=(x_test_padded, y_test), verbose=1)
```
